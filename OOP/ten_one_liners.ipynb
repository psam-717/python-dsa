# Reverse string

```python
text = "Pythone"
reversed_text = text[::-1]
#syntax
# string[start:end:step]
print(reversed_text)
```

# swapping two variables without a temp variable

```python
name, profession = "Marvin", "AI-engineer"
name, profession = profession, name
print(name, profession)
```

# checking if a word is a palindrome

```python
word = "madam"

if word==word[::-1]:
    print(f"The word '{word}' is a palindrome")
else:
    print(f"The word '{word}' is not a palindrome")
```

# count the frequency of letter in a list

```python
from collections import Counter
print(Counter(["a", "a", "b", "c", "d", "b", "b"]))
```

# get all even numbers from a list

```python
nums = [1, 3, 7, 10, 7, 4, 5, 9]
print([n for n in nums if n%2==0])
```

# flatten a list

```python
nested=[[1,2], [3,4],[5,6]]
flattened_list = [x for sub in nested for x in sub]
print(flattened_list)
```

# factorial of number

```python
import math

five_factorial = math.factorial(5)
print(five_factorial)
```

# common elements in a list

```python
num_list_one = [1,2,3,4,5]
num_list_two= [2,4,6,7,8]

common_elements = list(set(num_list_one) & set(num_list_two))
print(common_elements)
```

# union of elements in two list

```python
marks_one = [5,3,6,7,9]
marks_two = [1,2,3,10]

union = list(set(marks_one) | set(marks_two))
print(union)
```

# get unique elements and sort them

```python
list_one = [5,5,6,2,4,6]
sorted_unique_list = sorted(set(list_one))
# sorted() - sorts the elements in ascending order
# set() - removes duplicates
print(sorted_unique_list)
```

# one liner FizzBuzz

```python
# multiples of 3 will show Fizz, multiples of 5 will show Buzz and a multiple of 3 and 5 will show FizzBuzz
print(['Fizz'*(i%3==0)+'Buzz'*(i%5==0) or i for i in range(1,16)])
```
