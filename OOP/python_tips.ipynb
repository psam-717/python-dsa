# list comprehension

```python
squares = [x**2 for x in range(5)]
print(squares)
```

```python
names = ["apple", "fruit", "water-melon"]
for i, name in enumerate(names):
    print(i, name)
    
```

* merged dictionary

```python
dict1={"a": 1, "b": 2, "c": 3}
dict2={"d": 5, "e": 5}
merged={**dict1, **dict2}
print(merged)
```

* using _ instead of a variable in a for loop

```python
for _ in range(5):
    print("God is great")
```
