# Basic for loop

```python
for i in range(5):
    print(i)
```

# iterating through a list

```python
names= ["lisa", "abriel" , "lana", "ivanka", "vanessa"]
for i in names:
    print(i)
```

# loop with index

```python
cars = ["toyota", "honda", "hyundai", "benz", "maserati"]
for index, car in enumerate(cars):
    print(index, car)
```

# nested loops

```python
for i in range(2):
    for j in range (3):
        print(i, j)
```

# while loop

```python
count = 1
while count <=3:
    print("Count: ", count)
    count+=1
```

# while with a break statement

```python
num = 1
while True:
    if num > 3:
        break
    print("Number",num)
    num+=1
```

# continue in a loop

```python
for x in range(5):
    if x == 2:
        continue
    print(x)
# continue skips the current iteration
```

# looping through a dictionary

```python
user = {"name": "Kwame Apenteng", "age": 22, "occupation": "software engineer"}
for key, value in user.items():
    print(key,":", value)
```
