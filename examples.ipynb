```python
users = [
    dict(id=1, name="Marvin Sap", age=20, business_owner=True),
    dict(id=2, name="Sekoh Zap", age=10, business_owner=False),
    dict(id=3, name="phil", age=10, business_owner=False),
    dict(id=4, name="max", age=12, business_owner=True),
    dict(id=5, name="mike", age=15, business_owner=False),
    dict(id=6, name="psam", age=14, business_owner=True),
    dict(id=7, name="mphil", age=13, business_owner=True),
]
```

# first three users

```python
first_three = list(filter(lambda user: user['id'] < 4, users))
print(first_three) 
```

# business owners

```python
business_owners = list(filter(lambda user: user['business_owner'], users))
for owner in business_owners:
    print(owner['name'])
```
